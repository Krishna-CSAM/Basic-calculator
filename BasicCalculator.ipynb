```python
num_1 = int(input("first number"))
sign = input("math operation (+,-,*,/):")
num_2 = int(input("second number"))
if sign == "+":
   print(num_1 + num_2)
elif sign == "-":
  print(num_1 - num_2)
elif sign == "*":
   print(num_1 * num_2)
elif sign == "/":
  print(num_1/num_2)
else:
 print("error: invalid operation selected")
```
