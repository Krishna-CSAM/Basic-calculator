```python
import sympy as sp
x = sp.Symbol('x')
problem = input("what do you wanna do ?\n1.Calculas\n2.Trigonometric calculation\n3.logarithmic calculation\n4.Normal calculation\nEnter problem(1/2/3/4):")
if problem == "1":
  operation_1 = input("what operation?\n1.Integration\n2.Differentiation\nEnter operation_1(1/2):")
  function = input("what function?")
  if operation_1 == "1":
    answer = sp.integrate(function,x)
    print(answer)
  elif operation_1 == "2":
     answer = sp.diff(function,x)
     print(answer)
  else:
    print("error invalid choice, please choose between(1,2)")
elif problem == "2":
   Trigno_function = input("what is your trignometric function?\n1.Sin\n2.Cos\n3.Tan\n4.Cot\n5.Csc\n6.Sec\nEnter Trigno_function(1/2/3/4/5/6):")
   angle = int(input("angle in degrees"))
   rad = sp.rad(angle)
   if Trigno_function == "1":
    print("exact value:",sp.sin(rad))
    print("decimal value:",float(sp.sin(rad)))
   elif Trigno_function == "2":
     print("exact value:",sp.cos(rad))
     print("decimal value:",float(sp.cos(rad)))
   elif Trigno_function == "3":
    print("exact value:",sp.tan(rad))
    print("decimal value:",float(sp.tan(rad)))
   elif Trigno_function == "4":
    print("exact value:",sp.cot(rad))
    print("decimal value:",float(sp.cot(rad)))
   elif Trigno_function == "5":
    print("exact value:",sp.csc(rad))
    print("decimal value:",float(sp.csc(rad)))
   elif Trigno_function == "6":
    print("exact value:",sp.sec(rad))
    print("decimal value:",float(sp.sec(rad)))
   else:
      print("error:invalid choice, please choose between (1,2,3,4,5,6)")
elif problem == "3":
  log_type = input("which log?\n1.Natural log(ln - base e)\n2.Custom base log\nEnter log_type(1/2)")
  if log_type == "1":
    num = float(input("which number log has to be done?"))
    print("answer:",float(sp.log(num)))
  elif log_type == "2":
    num = float(input("which number log has to be done?"))
    base = float(input("base of log ?:"))
    print("answer:",float(sp.log(num, base)))
  else:
    print("error invalid choice please choose between(1,2)")
elif problem == "4":
  num_1 = float(input("first number:"))
  sign = input("math operation (+,-,*,/)")
  num_2 = float(input("second number:"))
  if sign == "+":
    print("answer:",num_1 + num_2)
  elif sign == "-":
    print("anwer:",num_1 - num_2)
  elif sign == "*":
    print("answer:",num_1*num_2)
  elif sign == "/":
    print ("answer:", num_1/num_2)
  else:
    print("error recheck your choices!")
else:
  print("error invalid choice please select between(1,2,3,4)")

```
